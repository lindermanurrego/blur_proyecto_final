```python

from pyspark.sql import functions as F
T_SILVER = 'blur_proyecto_final.silver.sales_silver'

df_ventas_cliente = (
    spark.table(f"{T_SILVER}")
    .groupBy("cliente")
    .agg(F.sum(F.col("monto")).alias("total_ventas"))
    .withColumn("total_ventas", F.round(F.col("total_ventas"), 2) )
    .orderBy(F.desc("total_ventas"))
    .limit(10)
)

df_ventas_cliente.write.mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable("blur_proyecto_final.gold.ventas_cliente")

display(df_ventas_cliente)
```

```python
df_cantidad_cliente = (
    spark.table(f"{T_SILVER}")
    .groupBy("cliente")
    .agg(F.sum(F.col("cantidad")).alias("cantidad_ventas"))
    .orderBy(F.desc("cantidad_ventas"))
    .limit(10)
)

df_cantidad_cliente.write.mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable("blur_proyecto_final.gold.cantidad_cliente")
display(df_cantidad_cliente)
```

```python
df_total_fecha = (
    spark.table(f"{T_SILVER}")
    .groupBy("date")
    .agg(F.sum(F.col("monto")).alias("total_ventas"))
    .withColumn("total_ventas", F.round(F.col("total_ventas"), 2) )
    .orderBy(F.desc("total_ventas"))
    .limit(10)
)

df_total_fecha.write.mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable("blur_proyecto_final.gold.total_fecha")
display(df_total_fecha)
```

```python
df_total_pais = (
    spark.table(f"{T_SILVER}")
    .groupBy("pais")
    .agg(F.sum(F.col("monto")).alias("total_ventas"))
    .withColumn("total_ventas", F.round(F.col("total_ventas"), 2) )
    .orderBy(F.desc("total_ventas"))
    .limit(10)
)

df_total_pais.write.mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable("blur_proyecto_final.gold.total_pais")
display(df_total_pais)
```
